## Obvezni del

Ves semester se ukvarjamo s sesalci, ki jim je potrebno podati pot. To je seveda iz mode: današnji sesalci samodejno bluzijo po prostoru, dokler ta ni posesan. Nekateri bolj, drugi manj uspešno.

Zadnji model sesalca podjetja Zapušek so videli delati takšne kroge.

Podrobnejši pregled je pokazal, da gre dejansko za kroge, postavljene okrog krožnice, tako:


Gre torej, v bistvu, za n krožnic, ki se "začenjajo" na n točkah enakomerno razporejenih okrog ("navidezne") manjše krožnice. Z drugimi besedami: narišete veliko krožnico, se premaknete za določen kot naprej po manjši notranji krožnici, ponovno narišete veliko krožnico in tako naprej.

Pa sprogramirajte želvo tako, da bo simulirala gibanje tega sesalca in narisala takšne slike.

## Rešitev

Tole je bolj razmišljanje o geometriji kot programiranje.

```python
import math
import zelva
import risar

def krog(r):
    n = 30
    a = 2 * math.pi * r / n
    for _ in range(n):
        ana.forward(a)
        ana.turn(360 / n)

r_not = 40
n_not = 12

ana = zelva.Turtle()
ana.pen_up()
ana.forward(r_not)
ana.pen_down()
ana.right()
for _ in range(n_not):
    krog(100)
    ana.forward(r_not * 2 * math.pi / n_not)
    ana.turn(360 / n_not)

risar.stoj()
```

Pripravili smo si funkcijo, ki nariše krog s podanim polmerom, ob čemer predpostavi, da je želva na eni od točk krožnice in obrnjena v smeri tangente.

`r_not` in `n_not` sta polmer notranje krožnice in število krogov. `n_not` nariše drugo sliko; za prvo, bi ga povečali z `12` na, recimo `120`.

Najprej zapeljemo želvo, kamor je treba: dvignemo pero, jo zapeljemo na rob notranje krožnice (`ana.forward(r_not)`) in obrnemo desno, saj mora biti obrnjena v smeri tangente. Nato `n_not`-krat narišemo zunanji krog (polmer smo fiksirali na 100), premaknemo želvo naprej za toliko, da bo vsota vseh premikov pri risanju notranjih krogov enaka obsegu notranje krožnice (njen obseg je `r_not * 2 * math.pi`) in jo obrnemo za toliko, da bo vsota vseh obratov enaka 360 stopinj.

## Neobvezna dodatna naloga

Tak model očitno ne posesa vse sobe. Boljše bi bilo, če bi sesalec opisal [Hilbertovo krivuljo](https://en.wikipedia.org/wiki/Hilbert_curve), vendar jo je malo sitno sprogramirati, predvsem z želvo.

Pa zato sprogramirajmo preprostjo [Zmajevo krivuljo](https://en.wikipedia.org/wiki/Dragon_curve).

- Zmajeva krivulja nivoja 0 je samo črta dolžine `a`.
- Desna zmajeva krivulja nivoja n je sestavljena iz **desne** zmajeve krivulje nivoja n-1, obrata **desno** in **leve** zmajeve krivulje nivoja n-1.
- Leva zmajeva krivulja nivoja n je sestavljena iz  **desne** zmajeve krivulje nivoja n-1, obrata **desno** in **leve** zmajeve krivulje nivoja n-1.

Pazite: obe varianti najprej vsebujeta desno, nato levo krivuljo nivoja n-1. Razlika je le v vmesnem obratu. (Da ne boste po nepotrebnem zapravili 15 minut, kot sem jih jaz. :)

Narišite torej to krivuljo. Priporočam, da napišete funkcijo `zmaj(nivo, smer)`, pri čemer je smer 1 za desno krivuljo in -1 za levo. Potem jo pokličite z
