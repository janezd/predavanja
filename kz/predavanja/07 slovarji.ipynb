# Slovarji

Zapiske o datotekah smo končali z groznim programom, ki se je trudil prešteti, kolikokrat je nekdo obiskal kateri hrib.

Datoteka je bila takšna:

Program smo organizirali tako, da je seznam `hribi` vseboval imena hribov, seznam `izletov` pa - na istoležnih elementih - število obiskov teh hribov.

Program je za vsak hrib preveril, ali je že v seznamu `hribi`. Če, potem je našel njegov indeks v `hribi` in povečal istoležni indeks v `izleti` za 1. Če ga še ni bilo, pa je v `hribi` dodal ime hriba, v `izleti` pa 1.

````python
hribi = []
izletov = []

for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib in hribi:
        for i, znani_hrib in enumerate(hribi):
            if znani_hrib == hrib:
                izletov[i] += 1
    else:
        hribi.append(hrib)
        izletov.append(1)

for hrib, izl in zip(hribi, izletov):
    print(hrib + ":", izl)
````

Ta program je kar zoprn. Kot bomo izvedeli prihodnjič, lahko iskanje indeksa precej poenostavimo, vendar ... še vedno je vse skupaj kar nerodno. Vendar: sledili smo ideji neke študentke in ideja je bila izvedljiva.

Malenkost preprosteje bi bilo, če bi delali z enim samim seznamom, ki bi kot elemente vseboval pare - sezname z imenom hriba in številom obiskov, recimo `[["Lubnik", 5], ["Grmada", 2], ...]`. (Tudi takšne pare bomo prihodnjič zapisovali drugače. Ampak za danes je to dovolj primerno.)

````python
izleti = []
for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    for par in izleti:
        if par[0] == hrib:
            par[1] += 1
            break
    else:
        izleti.append([hrib, 1])

for hrib, izl in izleti:
    print(hrib + ":", izl)
````

Zdaj gremo za vsak hrib prek parov. Če je prvi element para iskani hrib, povečamo drugi element in prekinemo zanko. Če zanke ne prekinemo (`else` po zanki -- poravnan je s `for` ne z `if`!), pa v seznam izletov dodamo nov par, namreč novi hrib in `1`.

Še vedno pa se moramo zafrkavati z iskanjem. To ni luštno.

## Kot množice, vendar s pari

Množice, kot smo videli, delujejo tako, da rezervirajo nekoliko preveliko tabelo (po Pythonovsko: seznam). Ko vanjo dodajamo elemente, za vsak element določijo, kam sodi in ga postavijo tja. Po tem sledi razreševanje *trkov*, ko je predvideno mesto že zasedeno, a to je stvar za druge predmete drugih študentov.

Tu lahko uporabimo podobno idejo. Sestavili bi lahko, recimo, seznam z 10 pari `[None, None]`.

````python
izleti = [[None, None]] * 10

izleti
````

Ko naletimo na določen hrib, recimo, Lubnik, pogledamo, kam v seznam sodi.

````python
hrib = "Lubnik"

mesto = hash(hrib) % 10
mesto
````

Če je Lubnik že tam (`if izleti[mesto][0] == hrib`), le povečamo drugi element tega para za `1` (`izleti[mesto][1] += 1`). Sicer pa ta par zamenjamo z novim hribom in `1` (`izleti[mesto] = [hrib, 1]`). Če je tam že kak drug hrib pa, kot rečeno, prepustimo v razmišljanje drugim.

Ker je ta stvar v resnici velikokrat uporabna, ima Python nekaj podobnega že vdelano. Pa drugi praktično uporabni jeziki večinoma tudi.

## Slovar

Slovar je križanec med seznamom in množico. Sestavimo ga, recimo, takole:

````python
visine = {"Triglav": 2864, "Storžič": 2132, "Šmarna Gora": 667}
````

Spet smo uporabili zavite oklepaje, le da tokrat v njem niso našteti le elementi, temveč pari. Prvemu elementu para (tule je to ime hriba) rečemo *ključ*, drugemu *vrednost*. Python je sestavil tabelo parov. Ključ določa, kje v tabeli je shranjen posamični element; vrednost mu je pripisana.

Slovar je podoben množici, ker na podoben način "razmetava" elemente in jih zato tudi hitro najde. Če nas zanima, ali je med slovarjevimi ključi Triglav, in ali je med njimi Grmada, odgovori v trenutku, saj točno ve, kam (če kam) ju je spravil.

````python
"Triglav" in visine
````

````python
"Grmada" in visine
````

Z operatorjem `in` lahko poizvedujemo le po ključih. Vrednosti ne gleda.

````python
2864 in visine
````

Razloga sta dva. Prvi je, da bi bilo zoprno, če bi gledal na oba konca, saj tako ne bi vedeli, kje je našel to reč. Kaj, če nas zanima le, ali se nekaj pojavi med ključi? Ali če se pojavi med vrednostmi? In, ja, navadno nas zanima le eno. Drugič: preveriti, ali obstaja določen ključ, je preprosto, saj so elementi razmetani po ključih. Da bi preveril, ali obstaja določena vrednost, pa bi moral pregledati vse vrednosti po vrsti. Slovarji so dobri pri delu s ključi, zato pač delajo to.

Hkrati pa je podoben seznamu. V seznamu lahko izvemo, kaj je shranjeno na določenem indeksu, v slovarju pa, kaj je shranjeno pri določenem ključu.

````python
visine["Triglav"]
````

````python
visine["Grmada"]
````

Ključ ima torej podobno vlogo kot indeks. Če ne obstaja dobimo napako `KeyError`.

Ker so elementi razmetani podobno kot pri množici, ne moremo govoriti o prvem, drugem ... zadnjem elementu. Zato številski indeksi ne delujejo. Če poskušamo s številskim indeksom dobiti, recimo, prvi element, bo Python mislil, da želimo vrednost, ki pripada ključu `1`.

````python
visine[1]
````

(Mimogrede: vrstni red elementov v slovarju je enak vrstnemu redu, v katerem jih dodajamo. Do tega je prišlo slučajno, nenamerno: malo so optimizirali način shranjevanja, pa so kot stranski učinek dobili takšen vrsten red.)

Novih elementov ne dodajamo z `append`, kot pri seznamih, saj bi to impliciralo dodajanje na konec, pa tudi ne z `add` kot pri množicah. Preprosteje je: le vrednost priredimo ključu.

````python
visine["Škrlatica"] = 2740
````

## Primer: izleti s slovarjem

Ponovno rešimo nalogo z izleti.

````python
izleti = {}
for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib not in izleti:
        izleti[hrib] = 0
    izleti[hrib] += 1

print(izleti)
````

Precej preprosteje, ni?

Za vsak hrib pogledamo, ali je v slovarju. Če ga še ni, ga dodamo, z vrednostjo 0. Nato v vsakem primeru povečamo število obiskov tega hriba -- če smo ga dodali pravkar, ga povečamo z `0` na `1`, če je bil tam že prej, pa pač s toliko, kot je bil prej.

Hribe bi bilo sicer potrebno lepše izpisati. Najbrž bo treba čez slovar z zanko. Hm, kaj pa dobimo, če gremo z zanko čez slovar? Morda bi bilo logično, da pare? Ja, bi bilo, po drugi strani pa bi bilo logično, da tole izpiše same "Da"-je.

````python
for x in izleti:
    if x in izleti:
        print("da")
````

Zgoraj smo že utemeljili, zakaj mora operator `in`, ko ga uporabljamo v pogoju (recimo `if`) delati s ključi. No, če zanka vrača tisto, kar je v slovarju, mora pogoj potem potrditi, da je to res v slovarju. Če se pogoj nanaša na ključe, je logično, da se bo tudi zanka. To ni nič hudega: če imamo ključ, nam je tako ali tako preprosto priti do pripadajoče vrednosti. Obratno bi bilo bolj zoprno ...

Pa izpišimo, končno, izlete.

````python
izleti = {}
for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib not in izleti:
        izleti[hrib] = 0
    izleti[hrib] += 1

for hrib in izleti:
    print(hrib + ":", izleti[hrib])
````

Obstaja tudi način, kako v zanki dobiti oboje. Ali pa celo vrednosti. A to naj nam ostane za prihodnjič.

## Rešitev domače naloge o izdatkih

V neki domači nalogi smo dobili datoteko z izdatki za hrano, bivanje in ostalo v obliki

```
H23
H5
B30
H8.25
O2
O4
B22
```

Naloga je bila napisati funkcijo, ki vrne seznam s tremi elementi: vsoto vseh številk pri H, pri B in pri O. Nalogo smo reševali s tremi spremenljivkami: zdaj jo znamo s slovarji.

````python
def izdatki(ime_datoteke):
    vsote = {}
    for vrstica in open(ime_datoteke):
        tip, znesek = razdeli(vrstica)
        if tip not in vsote:
            vsote[tip] = 0
        vsote[tip] += znesek
    return [vsote["H"], vsote["B"], vsote["O"]]
````

Je to boljše od treh spremenljivk? Seveda je: to je splošneje. Zdaj imamo lahko še deset drugih črk, ne da bi za to potrebovali še deset novih spremenljivk. Tudi funkcija bi lahko (če naloga ne bi zahtevala drugače) vrnila slovar namesto seznama.
