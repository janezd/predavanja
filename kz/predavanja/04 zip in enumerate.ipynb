## Ogrevanje in ponavljanje in preštevanje nizov

Za ogrevanje in ponavljanje se lotimo preproste naloge: preštejmo, koliko A-jev vsebuje ta ali oni izmed naslednjih nizov.

```python
a = "ATGGAGCGTGGTTGCGAGCGATAATCGACTAGCTAC"
b = "ACGUACAGCGAUCAGCUAACGAUGCAUGCAGGUACG"
c = "ACGUACAGCGTCAGCUAACGAUGCAUGCAGGUACG"
d = "AACGACGGGACAGCAGCGAGCAGCAGCCGACG"
```

Pač: štejemo. Potrebovali bomo spremenljivko, ki jo bomo povečali za 1 vsakič, ko v zanki prek niza naletimo na A.

```python
ajev = 0
for crka in a:
    if crka == "A":
        ajev += 1

ajev
```

(Mimogrede povejmo, da se kdo, ki ve že kaj več, ne bo pohujšal: A-je lahko preštejemo tudi z `a.count("A")`, vendar se tega, kar se skriva za tem, še nismo učili.)

Če `a` zamenjamo z `b`, bomo izračunalni število A-jev v drugem nizu.

Pa bi se kdo vprašal: koliko pa je A-jev v vseh štirih nizih skupaj. In nekdo drug je imel idejo: sestavimo seznam, ki vsebuje vse štiri nize. (Recimo jim *sekvence*, ker gre pač za genske sekvence.)

```python
sekvence = []
sekvence.append(a)
sekvence.append(b)
sekvence.append(c)
sekvence.append(d)
```

Ne, ne tako. No, lahko, vendar to ni potrebno. Tako kot smo seznam s par števili sestavili z `s = [5, 2, 6, 2, 1]`, ne pa tako, da smo pripravili prazen `s` in vanj eno za drugim dodali teh pet števil, bomo tudi seznam sekvenc sestavili z

```python
sekvence = [a, b, c, d]
```

To bi potem lahko nadaljevali z

```python
ajev = 0
for crka in sekvence:
    if crka == "A":
        ajev += 1

ajev
```

Očitno je nekaj narobe.

Problem je v tem: seznam `sekvence` ima štiri elemente, namreč sekvence a, b, c in d. Zanka `for` se bo torej obrnila štirikrat, vsakokrat priredivši po enega od teh štirih nizov spremenljivki `crka`.

```python
for crka in sekvence:
    print(crka)
```

Noben od teh štirih nizov ni `"A"` torej imamo 0 `ajev`.

V resnici bi se morala zanka glasiti `for sekvenca in sekvence`. In znotraj tega gremo čez črke (poslej jim recimo baze, ker ... genetika) vseh sekvenc, torej

```python
sekvence = [a, b, c, d]

ajev = 0
for sekvenca in sekvence:
    for baza in sekvenca:
        if baza == "A":
            ajev += 1

ajev
```

Seznama `sekvence` nam ni potrebno sestavljati vnaprej, sploh, če ga potrebujemo le zato, da prek njega naženemo zanko. Lahko pišemo kar

```python
ajev = 0
for sekvenca in [a, b, c, d]:
    for baza in sekvenca:
        if baza == "A":
            ajev += 1

ajev
```

Zanko znotraj zanke bomo uporabili, kadar imamo sezname znotraj seznamov ali, v tem primeru, nize znotraj seznamov. Vse skupaj si lahko predstavljamo kot nekakšno (dvodimenzionalno) tabelo: prva zanka gre prek vrstic, notranja pa prek stolpcev vsake od njih. (Pri tem ni nujno, da ima vsaka vrstica enako "stolpcev"; v našem primeru že ni tako, saj so nizi različno dolgi.)

Gre pa tudi preprosteje: če nas zanima le skupno število a-jev, zlepimo nize skupaj, pa imamo eno samo zanko.

```python
ajev = 0

for baza in a + b + c + d:
    if baza == "A":
        ajev += 1

ajev
```

## Znanka prek dveh seznamov (ali nizov) vzporedno

Zdaj pa vzemimo dva niza,

```python
a = 'ATGGAGCGTGGTTGCGAGCGATAATCGACTAGCTAC'
b = 'ATGGAGTGTGAAAAAAAAAAATAAGGGACTAGCTAC'
```

Ugotoviti želimo, v koliko znakih se ujemata. (Genetike take reči kar pogosto zanimajo.)

Očitno potrebujemo zanko čez prvi niz, `for baza_a in a`. Vsak `baza_a` bi radi primerjali z *istoležnim* elementom iz niza `b`. Seveda nas zamika napisati

```python
ujemanj = 0

for baza_a in a:
    for baza_b in b:
        if baza_a ==  baza_b:
            ujemanj += 1

ujemanj
```

Očitno čisto preveč.

Kaj se dogaja, poglejmo na tem primeru:

```python
punce = ["Ana", "Berta", "Cilka"]
fantje = ["Anže", "Benjamin", "Cene"]

for punca in punce:
    for fant in fantje:
        print(punca, "pleše z", fant)
```

Zunanja zanka gre čez vse punce in *za vsako punco* čez vse fante. Enako se je zgodilo pri genskih sekvencah: vsak element prve je primerjal z vsemi elementi druge. Že prvi `A` iz `a` se je uje(ma)l z **vsemi** `A` je iz `b`. Odtod 339 ujemanj.

**Medklic:** ker sem zgoraj izdal, da lahko `A`-je v `a` preštejemo kar z `a.count("A")` in tako naprej, se lahko hitro prepričamo, da je tule res poparil vsak `A` iz `a` z vsemi `A`-ji iz `b`, `C`-je s `C`-ji, `G`-je z `G`-ji in `T`-je s `T`-ji. 

```python
a.count("A") * b.count("A") + a.count("C") * b.count("C") + a.count("G") * b.count("G") + a.count("T") * b.count("T")
```

In ker se poznamo zanke, lahko to preštejemo tudi hitreje.

```python
pari = 0
for baza in "ACGT":
    pari += a.count(baza) * b.count(baza)

pari
```

In kako nam je žal, da se pri tem predmetu najbrž ne bomo naučili narediti tega še hitreje!

```python
sum(a.count(baza) * b.count(baza) for baza in "ACGT")
```

**Konec medklica.**

Želeli bi torej zanko prek obeh hkrati, nekaj v slogu

```python
ujemanj = 0

for baza_a in a:
for baza_b in b:
    if baza_a == baza_b:
        ujemanj += 1
```

To očitno ne dela.

Po pravici, tega še ne znamo. Pa se naučimo!

Pomagali si bomo z odlično funkcijo `zip`. Kot argument ji damo dve (ali tri, štiri ali več stvari), čez kakršne navadno spuščamo zanke, se pravi dva (tri ...) sezname ali nize. Vrnila bo seznam parov (trojk, četverk...) istoležnih elementov podanih stvari.

```python
zip(punce, fantje)
```

Hm, je tole seznam? No, ni. Funkcija vrne nekaj, kar se (v določenih pogledih) vede kot seznam. Za potrebe, ki jih imamo tule, se lahko vedemo, da vrne seznam, namreč takšen

```python
list(zip(punce, fantje))
```

Če se delamo, da nisem napisal tistega `list`, nam `zip(punce, fantje)` torej vrne seznam parov -- oba prva, oba druga, oba tretja elementa.

Podamo mu lahko tudi več seznamo, na primer

```python
list(zip(punce, fantje, [5, 2, 6, 6, 0]))
```

Ali pa tudi

```python
list(zip("Berta", "Beno"))
```

In celo

```python
list(zip(punce, fantje, "ACGT"))
```

Mimogrede smo opazili tudi, da se `zip` se ustavi, ko zmanjka najkrajše od podanih stvari, recimo črk v nizu Beno ali v zadnjem primeru, punc in fantov.

## Znanka prek dveh seznamov (ali nizov) vzporedno [nadaljevanje]

Zdaj znamo dovolj, da preverimo število ujemanj. No, najprej organizirajmo tisti ples, ne da bi vsaka punca morala plesati z vsemi fanti po vrsti.

```python
for par in zip(punce, fantje):
    print(par)
```

Tule smo se samo prepričali, da `zip` res dela, kar smo obljubili. Zdaj pa razpakirajmo par in ga izpišimo.

```python
for par in zip(punce, fantje):
    punca, fant = par
    print(punca, "pleše z", fant)
```

Preden gremo na baze, pokažimo še bližnjico. Ta vzorec je v resnici pogost -- namreč zanka čez nek zip in nato razpakiranje. Oboje lahko postorimo v enem koraku: spremenljivke par niti ne potrebujemo, razpakiranje izvede že kar `for`:

```python
for punca, fant in zip(punce, fantje):
    print(punca, "pleše z", fant)
```

Zdaj pa le k bazam.

```python
a = 'ATGGAGCGTGGTTGCGAGCGATAATCGACTAGCTAC'
b = 'ATGGAGTGTGAAAAAAAAAAATAAGGGACTAGCTAC'

ujemanj = 0

for baza_a, baza_b in zip(a, b):
    if baza_a == baza_b:
        ujemanj += 1

ujemanj
```

(Mimogrede, kakšna škoda je, da ne bomo znali napisati tega!

```python
sum(baza_a == baza_b for baza_a, baza_b in zip(a, b))
```

)

Pa se je neka študentka vprašala: bi lahko izpisala, na katerih mestih je prišlo do razlik? Recimo zaporedne številke črk?

S tem, kar že znamo, lahko to naredimo tako:

```python
ujemanj = 0
pozicija = 1

for baza_a, baza_b in zip(a, b):
    if baza_a == baza_b:
        ujemanj += 1
    else:
        print(pozicija)
    pozicija += 1

print("Število ujemanj:", ujemanj)
```

V spremenljivki pozicija preprosto štejemo: v začetku je `1`, potem pa jo v vsakem krogu zanke povečamo za `1`.

Obstajajo pa tudi elegantnejši načini. Ti si zaslužijo ločen razdelek.

## Oštevilčeni seznami in neskočno štetje

Za začetek nekaj preprostejšega: izpišimo lokacije vseh `T`-jev v nizu `a`.

Po prejšnji metodi, storimo tako.

```python
pozicija = 1
for baza in a:
    if baza == "T":
        print(pozicija)
    pozicija += 1
```

Tole, namreč element seznama in njegovo zaporedno številko (učeno: njegov indeks) potrebujemo kar pogosto. Zato obstaja funkcija, ki je (malo) podobna `zip`. `zip` prejme dva (ali več) seznamov in vrne pare (ali večterke). Funkcija `enumerate` pa zahteva en sam seznam (ali niz) in vrne "seznam" parov (indeks, element).

```python
enumerate(punce)
```

Ah. Spet v resnici ne vrne seznama temveč nekaj, kar se (v mnogo pogledih) vede podobno kot seznam. Namreč kot takšen seznam:

```python
list(enumerate(punce))
```

Enumerate torej šteje ("enumerira") namesto nas. Pa izpišimo položaje `T`-jev z `enumerate`.

```python
for pozicija, baza in enumerate(a):
    if baza == "T":
        print(pozicija)
```

Dobra novica: to je preprosteje, saj vrednosti `pozicija` ne šofiramo mi, temveč jo `enumerate`. Slaba novica: vse številke so za `1` premajhne. Kot tudi ni ušlo budnemu očesu študenta, študentke, ki je brala prejšnji primer: Ani je bila dodeljena številka `0` namesto `1`.

Tudi to je ena imenitnih navad računalnikarjev: šteti začnemo z `0`, ne z `1`. Se pa pustimo prepričati, če je treba: funkcija `enumerate` sprejema dodaten argument, ki ga je vedno potreba podajati *poimensko*, takole:

```python
list(enumerate(punce, start=1))
```

Ali, če je potrebno

```python
list(enumerate(punce, start=42))
```

Zdaj pa k našemu prvotnemu cilju: zanima nas, kje so razlike med zaporedjema `a` in `b`. Spet ju bomo zipnili skupaj, da bomo lahko primerjali istoležne elemente, potem poleg tega pa te pare še enumerirali.

```python
for pozicija, baza_a, baza_b in enumerate(zip(a, b)):
    if baza_a != baza_b:
        print(pozicija)
```

Zakaj pa ne? Napaka se je zgodila v `for`: pritožuje se, da zahtevamo razpakiranje v tri vrednosti (`pozicija`, `baza_a` in `baza_b`), medtem ko je `enumerate(zip(a, b))` priskrbel le dve. Kako? Mar ni `zip` naredil parov, `enumerate` pa prilepil k temu še številk? Je, vendar ... številke je prilepil k parom, ne pa dodal k paru (in s tem naredil trojke).

Tole poglejmo na puncah in fantih, da bo izpis krajši.

```python
list(enumerate(zip(punce, fantje)))
```

Elementi "seznama", ki ga dobimo z `enumerate` so vedno pari, sestavljeni iz številk in elementov tega, kar smo dali enumerirati. Enumerirati smo dali pare, torej dobimo pare, sestavljene iz številk in parov.

Uh, potemtakem ...

```python
for pozicija, par_baz in enumerate(zip(a, b)):
    baza_a, baza_b = par_baz
    if baza_a != baza_b:
        print(pozicija)
```

Lahko, vendar gre tudi preprosteje (ali zapleteneje, kakor pogledamo na stvar). Tudi par baz lahko razpakiramo že v glavi zanke, vendar mora razpakiranje odražati strukturo, ki jo razpakira. Napisati moramo tako:

```python
for pozicija, (baza_a, baza_b) in enumerate(zip(a, b), start=1):
    if baza_a != baza_b:
        print(pozicija)
```

Če bo komu lažje razumeti: v bistu delamo tole

```python
for (pozicija, (baza_a, baza_b)) in enumerate(zip(a, b), start=1):
    if baza_a != baza_b:
        print(pozicija)
```

... saj smo dobili par, ki vsebuje številko in par. Vendar smemo izpustiti zunanje oklepaje in tako dobimo prejšnji zapis.

Če hočemo preprosteje (ali, spet, morda bolj zapleteno), lahko uporabimo nenavadno funkcijo `count`, ki vrne seznam (oziroma, spet, nekaj, kar se vede, kot da bi bilo seznam) števil od 0 do neskočno. Funkcijo pričaramo z `from itertools import *`. Neskončnega "seznama", ki ga sestavi `count`, ne bomo izpisovali (ker bi trajalo predolgo), pač pa ga bomo uporabili v `zip`.

```python
from itertools import *

list(zip(count(), punce, fantje))
```

`zip` zdaj sestavlja trojke, ki vsebujejo, kar priskrbijo `count()` ter seznama `punce` in `fantje`. Ker seznamov s puncami in fanti zmanjka neprimerno prej kot neskončnega seznama števil, `zip` hitro zaključi svoje delo.

Tudi `count` lahko podamo številko, pri kateri naj začne štetje.

```python
from itertools import *

list(zip(count(1), punce, fantje))
```

Mesta, na katerih se razlikujejo baze, torej lahko izpišemo tudi z

```python
for pozicija, baza_a, baza_b in zip(count(1), a, b):
    if baza_a != baza_b:
        print(pozicija)
```
