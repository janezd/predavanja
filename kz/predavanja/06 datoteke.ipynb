## Branje datotek

Datoteko odpremo tako, da pokličemo funkcijo `open`, kot argument ji podamo ime datoteke.

```python
open("izleti.txt")
```

Hm, kaj je pravzaprav vrnila ta funkcija? In kje se je odprla datoteka?

Če bi se datoteka "odprla" v smislu, da bi se nekje odprlo okno z njeno vsebino, to ne bi bilo prav uporabno. Sama beseda "odpiranje" je nenavadna. Datoteka se je "odprla" tako, da je Python (oziroma operacijski sistem, ampak ne rinimo v globino) preveril, ali datoteka obstaja in ali jo imamo pravico brati ter sestavil "objekt" (navadno na tablo rišemo neke škatlice), ki predstavlja datoteko. V tej škatlici piše, na katero datoteko (očitno na *izleti.txt*) se nanaša, do kod jo je že prebral in še nekaj stvari.

Datoteko lahko, tako kot vse drugo, shranimo tudi v spremenljivko.

```python
f = open("izleti.txt")
```

```python
f
```

In kaj zdaj? Kaj lahko počnemo s takšno spremenljivko? Ob vsaki reči, ki smo jo spoznali doslej, smo izvedeli, kaj omogoča. Števila lahko seštevamo in odštevamo, sezname pa le seštevamo. Poleg tega smo lahko šli čez sezname z zanko ...  Lahko seštevamo datoteke, jih množimo, množimo s števili?

```python
f + f
```

```python
f * f
```

```python
2 * f
```

Nič od tega. Pač pa lahko gremo čez datoteko z zanko!

```python
for vrstica in f:
    print(vrstica)
```

Še enkrat poglejmo zadnjo vrstico.

```python
print(vrstica)
```

Ne, ne tako, brez `print`.

```python
vrstica
```

Tu vidimo nekaj zanimivega. Kaj je ta `\n`?

`\n` je znak za konec vrstice (ali začetek nove, kakor hočemo razumeti). Ozadje je razloženo v drugi strani z zapiski v istem poglavju učilnice. Tule nam je pomembno, da je `\n` predstavitev enega samega znaka. O tem se lahko hitro prepričamo:

```python
len(vrstica)
```

Sedem znakov, to je `L`, `u`, `b`, `n`, `i`, `k` in `\n`. Vprašamo lahko tudi po zadnjem znaku:

```python
vrstica[-1]
```

In dolg je le 1:

```python
len(vrstica[-1])
```

Več, kot rečeno, v ločenih zapiskih. Tu nas zanima le, kako se ga znebiti. Kmalu bomo spoznali boljši način, danes pa bomo preprosto odbili zadnji znak:

```python
hrib = vrstica[:-1]
hrib
```

Ponovimo izpisovanje.

```python
for vrstica in f:
    hrib = vrstica[:-1]
    print(hrib)
```

Hm, kaj smo ga polomili? Nič, samo datoteka je že porabljena. Vsaka vrstica, ki jo preberemo, je prebrana in je ne bomo dobili še enkrat. Datoteko moramo odpreti ponovno.

```python
f = open("izleti.txt")
for vrstica in f:
    hrib = vrstica[:-1]
    print(hrib)
```

Ali pa jo, še lažje, odpremo kar v glavi zanke, namesto da bi si zapletali življenje z dodatno spremenljivko.

```python
for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    print(hrib)
```

Izpis je zdaj lepši. Prej je vseboval kup praznih vrstic -- in zdaj vemo tudi, zakaj. Na koncu vsake vrstice je bil `\n`, zaradi katerega je šel `print` v novo vrstico, potem pa je, kot mu je sicer v navadi šel (še enkrat) v novo vrstico. Tako smo dobili prazne vrstice.

## Štetje

Pa preverimo: kolikokrat je bil možakar, ki je sestavljal to datoteko, na Lubniku?

```python
lubnik = 0
for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib == "Lubnik":
        lubnik += 1

print(lubnik)
```

## Ena grozna naloga

Zdaj pa si zastavimo drugačno nalogo: radi bi izpisali, kolikokrat je bil na vsakem hribu.

```python
lubnik = 0
grmada = 0
porezen = 0

for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib == "Lubnik":
        lubnik += 1
    if hrib == "Grmada":
        grmada += 1
    if hrib == "Porezen":
        porezen += 1

print("Lubnik:", lubnik)
print("Grmada:", grmada)
print("Porezen:", porezen)
```

Tole je grozno, pa še pomanjkljivo. Za začetek: kaj pa Ratitovec, Krim in Triglav? Še tri spremenljivke, še trije `if`-i s pripadajočo vrstico v njih in še trije `print`-i. Kaj pa, ko bo šel še na Vrtaško sleme, Stol, Jalovec in Rodico? Še štiri spremenljivke in tako naprej. Predvsem pa: kako napisati takšen program, če sploh ne vemo vnaprej, na katerega hribe ga bo še nosilo?

Ko smo na predavanjih prišli do te točke, nam je ostalo le še 15 minut, pa sem na hitro sprogramiral nekaj, poskušal razložiti in kljub temu, da smo (spet) zavlekli predavanja čez peto, večina programa najbrž ni najbolj razumela, predvsem pa je bila zgrožena. (Če bom te zapiske uporabil še kako prihodnje leto, pač vedite, da se je tako zgodilo neki generaciji pred vami.)

```python
hribi = []
izletov = []

for vrstica in open("izleti.txt"):
    hrib = vrstica[:-1]
    if hrib in hribi:
        for i, znani_hrib in enumerate(hribi):
            if znani_hrib == hrib:
                izletov[i] += 1
    else:
        hribi.append(hrib)
        izletov.append(1)

for hrib, izl in zip(hribi, izletov):
    print(hrib + ":", izl)
```

Komur se poglablja, naj se poglobi. Ni čisto brez zveze; celo nekoliko poučno je. Vendar smo se strinjali, da se tako ne da programirati. Kar je odlično izhodišče za to, kar se bomo šli prihodnji teden.
