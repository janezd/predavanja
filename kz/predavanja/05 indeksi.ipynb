## Naključni dostop

V starih filmih vidimo stare računalnike, ki imajo nekakšne kolute s trakovi. To je bilo še pred SSD diski, ki zapisujejo podatke na čipe (in nimajo gibljivih delov) in celo pred "klasičnimi" [trdimi diski](https://en.wikipedia.org/wiki/Hard_disk_drive) in [disketami](https://en.wikipedia.org/wiki/Floppy_disk), ki so zapisovali informacije na magnetno površino v obliki ... no, diska, okrogle plastične ploskve z magnetnim premazom. Podatki na [tračnih enotah](https://en.wikipedia.org/wiki/Magnetic_tape) so posneti na magnetnem traku, ki ga je potrebno pisati in brati tako, da vrtimo trak, navit na kolutih, mimo bralne enote.

Lahko imamo smolo: prvi podatek je na začetku traku, drugi na koncu, tretji nekje na sredi ... V tem primeru bo potrebno vrteti trak naprej in nazaj in naprej. Za takšno dostopanje do podatkov je tračna enota kar neprimerna reč. V splošnem so bile tračne enote najbolj primerne za branje podatkov po vrsti.

Računalnik potrebuje tudi drugačno vrsto pomnilnika, takšno, ki omogoča, da brez težav beremo podatke iz različnih mest, ne da bi zaradi navidez naključnega izbiranje mest (malo na začetku, malo na koncu ...) preveč trpela hitrost. Takšnemu pomnilniku, ki omogoča naključne dostope, so nadeli ime *pomnilnik z naključnim dostop* ali *random access memory*, oziroma, s kratico, *RAM*.

## Seznami omogočajo naključen dostop

Kako bi v naših seznamih izpisali vrednost, recimo, četrtega elementa? Pripravimo tabelo.

````python
imena = ["Ana", "Berta", "Cilka", "Dani", "Ema", "Fanči", "Greta"]
````

In zdaj izpišimo četrto ime.

````python
for i, ime in enumerate(imena, start=1):
    if i == 4:
        print(ime)
````

Tole je ekvivalentno prevrtavanju magnetnega traku: celo tabelo smo "prevrteli" skozi zanko in ko smo se nahajali na četrtem elementu, smo izpisali njegovo vrednost. Zanko bi lahko tudi prekinili,

````python
for i, ime in enumerate(imena, start=1):
    if i == 4:
        print(ime)
        break
````

## Rezine

Kaj pa bi lahko pomenilo tole:

```
imena[2:5]
```

Naj izdam: vsa imena od drugega do petega. Koliko imen bi bilo to? Lepo bi bilo, če bi bila tri, ker je 5 - 2 pač tri. Če gremo na službeno pot od 17. do 22., bi bilo lepo, če bi bili na poti pet dni, ker je 22 - 17 = 5, in ne 6 (o čemer se itak vedno prepričamo tako, da to še preštejemo na prste, za zihr).

Pa se dogovorimo, da bo `2:5` pomenilo tri elemente. Svet bo na ta način preprostejši. Za to bo sicer potrebno požreti še nekaj nenavadnega: `2:5` bo vrnil drugi, tretji in četrti element, ne pa tudi petega. Interval 2:5 torej vključuje element z indeksom 2, ne pa tudi tistega z indeksom 5.

````python
imena
````

````python
imena[2:5]
````

Če hočemo prva tri imena, torej napišemo

````python
imena[0:3]
````

In če hočemo vsa imena od tretjega do konca, pišemo

````python
imena[3:len(imena)]
````

Zdaj pa nas čaka nekaj imenitnih reči: spodnjo (ali zgornjo) mejo lahko izpustimo, če gremo od začetka (ali konca).

````python
imena[:3]
````

````python
imena[3:]
````

Tole se v resnici bere imenitno.

- `imena[:3]` pomeni **prva tri imena**, namreč tista z indeksi 0, 1 in 2 (ne pa 3, ker zgornja meja ni vključena).
- `imena[3:]` pomeni **vsa imena, razen prvih treh**, saj začne pri elementu z indeksom 3, izpustivši tiste z 0, 1 in 2.

A to še ni vse. Indeksi so, vemo, lahko tudi negativni.

- `imena[-3:]` pomeni **zadnja tri imena**, namreč imena z indeksi -3, -2 in -1.
- `imena[:-3]` pomeni **vsa imena, razen zadnjih treh**, saj zgornja meja ni vključena, zato manjka tudi element z indeksom -3 (poleg njega pa seveda še -2 in -1).

To je kar elegantno: ko imamo v indeksu 3 (ali -3), to vedno pomeni nekaj v zvezi s prvimi tremi (ali zadnjimi tremi) elementi. Ne štirimi, ne dvema: tremi. Ta eleganca je posledica sprejetih dogovorov, namreč, da štejemo z leve od 0, z desne od -1 (-0 očitno ne bi delovalo :), in da interval ne vključuje zgornje meje.

Računalnikarji smo vseeno praktični ljudje.

Pa še to: tej stvari, 2:5, rečemo *rezina*, po angleškem *slice*.

Še en zanimiv primer: takole seznamu obtešemo prvi in zadnji element.

````python
imena[1:-1]
````

Izpustimo lahko tudi obe meji,

````python
imena[:]
````

Kakšen smisel ima to? Ima. Rezinam lahko namreč podamo tudi *korak*. Za začetek, na primer, takole.

````python
imena[1:-1:2]
````

Vzeli smo elemente vse od prvega do zadnjega (torej brez ničtega in zadnjega), vendar vsakega drugega.

Ali, če hočemo, vsakega tretjega.

````python
imena[1:-1:3]
````

Tule smo izpustili prvi in zadnji element. Če ju hočemo obdržati, pač ne določimo spodnje in zgornje meje, temveč le korak.

````python
imena[::2]
````

Tole vrne ničti, drugi, četrti, šesti element seznama. Če bi hoteli tiste z lihimi indeksi, pač začnemo pri prvem.

````python
imena[1::2]
````

Če bi se torej hotela Ana pogovarjati z Berto, Cilka z Dani in Ema s Fanči, zipnemo skupaj tadva seznama.

````python
for prva, druga in zip(imena[::2], imena[1::2]):
    print(prva, "se pogovarja z", druga)
````

Morda vse te ženske sedijo v vrsti, pa bi radi napisali, da Ana sedi levo od Berta, Berta levo od Cilke in tako naprej? V tem primeru moramo popariti tadva seznama:

````python
print(imena)
print(imena[1:])
````

Drži? Prvi par bo Ana in Berta, drugi Berta in Cilka in tako naprej do Fanči in Greta, kjer se bo `zip` ustavil.

````python
for prva, druga in zip(imena, imena[1:]):
    print(prva, "je levo od", druga)
````

Se spomnimo genskih sekvenc?

````python
a = "ATGGAGCGTGGTTGCGAGCGATAATCGACTAGCTAC"
````

Genetiki radi gledajo po tri baze skupaj; temu rečejo kodon (če se prav spomnim). Torej 

```
ATG
GAG
CGT
GGT
TGC
```

in tako naprej. Znamo izpisati kodone v gornjem A?

Trik je v tem: v prvem stolpcu gornjega izpisa so AGCGT; to so baze na indeksih 0, 3, 6 ... V drugem stolpcu imamo TAGGG, ki jih najdemo na indeksih 1, 4, 7 ... V tretjem pa so GGTTC, ki so na indeksih 2, 5, 8 ...

Z drugimi besedami (torej: z besedami v Pythonu) so v prvem stolpcu

````python
a[::3]
````

v drugem

````python
a[1::3]
````

in v tretjem

````python
a[2::3]
````

Te tri je treba pozipati skupaj, da dobimo najprej ATG, nato GAG ... in jih izpisati.

````python
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    print(d, e, f)
````

Skoraj, samo še presledki nas motijo. Ni problema: samo seštejmo tele `d`, `e` in `f`:

````python
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    print(d + e + f)
````

Ali pa, malo postopneje,

````python
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    kodon = d + e + f
    print(kodon)
````

Pa se je našla študentka (vedno se najde študentka), ki jo je zanimalo, če lahko iz teh trojk spet sestavimo prvotni niz.

Seveda, samo sešteti jih moramo. Pripravimo si prazen niz in vanj dodajajmo vse, kar izpisujemo:

````python
prvotni_niz = ""
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    kodon = d + e + f
    print(kodon)
    prvotni_niz += kodon

print("Prvotni niz:", prvotni_niz)
````

Da bo lepše, dodajmo med kodone še ločilo, `|`. (Pa izpisovanje posameznih kodonov izpustimo, saj le podaljšuje zapiske.)

````python
prvotni_niz = ""
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    kodon = d + e + f
    prvotni_niz += kodon + "|"

print(prvotni_niz)
````

To je kar imenitno, le nekoga, ki ima dovolj OCD, moti `|` na koncu. Kako se ga znebimo?

En način je, da ga sploh ne dodamo. Pravzaprav to ni le en način; načinov, na katere ga lahko ne dodamo, je veliko in najimenitnejših med njimi žal najbrž ne bomo spoznali.

````python
"|".join(d + e + f for d, e, f in zip(a[::3], a[1::3], a[2::3]))
````

Ampak enega pa pravzaprav že poznamo: nizu `prvotni_niz` preprosto odbijemo zadnji znak:

````python
prvotni_niz = ""
for d, e, f in zip(a[::3], a[1::3], a[2::3]):
    kodon = d + e + f
    prvotni_niz += kodon + "|"

print(prvotni_niz[:-1])
````
