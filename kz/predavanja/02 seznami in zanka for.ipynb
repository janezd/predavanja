### Seznami

Na prejšnjih predavanjih smo iskali največje število v Collatzovem zaporedju, ki se začne z nekim številom, na primer 23.

````python
x = 23
naj = x
while x > 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    if x > naj:
        naj = x

print(naj)
````

In še druge podobne stvari smo počeli, recimo izpisali vsoto vse teh števil.

Pa se je nekomu zazdelo, da bi morala za to obstajati neka funkcija, ki bi ji preprosto podali "te številke", pa bi vrnila največjo.

Ideja je dobra, tudi imena funkcij, ki vrneta največje število in vsoto število povem -- imenujeta se `max` in `sum`. Hudič pa se skriva v detajlu: kaj naj bi podali tema funkcijama? "Te številke"?

````python
x = 23

while x > 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    te_stevilke = x

print(max(te_stevilke))
````

Seveda ne. `te_stevilke` bo vedno zadnja številka in ne "vse številke", ki so se pojavile v zaporedju. Še več, zadnja številka bo vedno `1`, torej gornji program ne izpiše drugega, kot bi izpisal

````python
max(1)
````

In, no ja, prav to isto tudi izpiše. `max` pričakuje več številk, ne ene same.

Doslej poznamo `int`, `float` in `bool` (no, in `str`, a ta je v pogledu današnje razprave malo poseben). Vsi ti vsebujejo eno samo število oziroma eno samo vrednost `True` ali `False`. Kar potrebujemo, je cel seznam števil. No, takole ga naredimo.

### Seznam števil v Collatzovem zaporedju

Zdaj znamo dovolj, da si pripravimo seznam vseh števil v Collatzovem zaporedju, ki se začne s 23.

Spomnimo se: takole bi jih izpisali.

````python
x = 23

while x != 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    print(x)
````

Nam pa ni do izpisovanja, temveč jih želimo zložiti v seznam. Poimenujmo ga `stevila`. Pred zanko ga seveda pripravimo (`stevila = []`), v zanki pa `print(x)` zamenjamo s `stevila.append(x)`.

````python
x = 23
stevila = []
while x != 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    stevila.append(x)
````

Izpisalo se ni nič, `stevila` pa vsebuje vse, kar smo zložili vanj.

````python
stevila
````

Prvo število, 23, manjka. No, v resnici ga tudi v prejšnjem programu nismo izpisali. (Kako to, da ne? Tako: `print` je bil šele na koncu zanke, po tem, ko smo iz prvega števila že izračunali naslednje.) To lahko uredimo tako, da pred zanko v seznam dodamo prvo število.

````python
x = 23
stevila = []
stevila.append(x)
while x != 1:
    if x % 2 == 0:
        x = x // 2
    else:
        x = 3 * x + 1
    stevila.append(x)

stevila
````

### Zanka for

Kako bi izpisali vsa liha števila iz seznama `stevila`?

V bistvu bi potrebovali tole

```
za vsako število x iz seznama števila ponovi tole:
    if x % 2 == 1:
        print(x)
```

Ko nekaj ponavljamo, je to zanka. Vendar tole ne diši po `while`: `while` ima nek pogoj in se ponavlja, dokler je pogoj resničen. Ta zanka pa je drugačne narave, ponavlja se "za vsak element seznama".

Takole nam je napisati:

````python
for x in stevila:
    if x % 2 == 1:
        print(x)
````

Če smo stari zanki, znanki prejšnjega tedna, rekli *zanka while*, bomo tej *zanka for*. Če imamo torej:

```
for x in stevila:
    <en kos programa>
```

bo šla zanka prek elementov seznama `stevila`. V vsakem krogu bo vzela en element seznama, ga priredila imenu `x` (seveda lahko uporabimo tudi poljubno drugo ime, ne nujno `x`) in izvedla `<en kos programa`, recimo gornji vrstici).

Nekatere študente zmoti, da v tem programu nikjer ne piše `x = ...`. Natančno tako: vrednosti `x` ne določamo mi, temveč ga določa zanka. In to vedno znova.

Če so števila takšna:

````python
stevila
````

je program

````python
for x in stevila:
    if x % 2 == 1:
        print(x)
````

ekvivalenten temu:

````python
x = 23
if x % 2 == 1:
    print(x)

x = 70
if x % 2 == 1:
    print(x)

x = 35
if x % 2 == 1:
    print(x)

x = 106
if x % 2 == 1:
    print(x)

# ... in tako naprej
````

### Težja vaja: Prvo sodo število (če je kakšno)

Pripravimo si malo krajši seznam.

````python
s = [5, 8, 3, 12, 4, 5]
````

Naša naloga bodi izpisati prvo sodo število.

Zamislimo si nalogo, v kateri nekdo izgovarja števila, drugi pa mora le ponoviti prvo sodo število (morebitna nadaljnja števila pa, tako kot liha, zamolčati). Poslušal bo števila in *čakal*. Ko pride prvo sodo, ga ponovi in poslej ne čaka več.

Kadarkoli se gremo takšne "miselne poskuse" velja: to, kar moramo vedeti, to kar si moramo zapomniti, si bo moral zapomniti tudi računalnik. Potrebovali bomo spremenljivko, v kateri bo zabeleženo, ali čakamo ali pa smo že dočakali.

````python
čakam = "da"

for x in s:
    if čakam == "da" and x % 2 == 0:
        print(x)
        čakam = "dočakal"
````

Zastavico (kot običajno rečemo takim spremenljivkam) `čakam` smo v začetku nastavili na "čakam". V zanki preverimo, ali še čakamo sodo število in če ga, preverimo, ali je `x` sod. Če je tako, ga izpišemo in postavimo `čakam` na `"dočakal"`. Tako bo v naslednjih krogih zanke pogoj `čakam == "da" and x % 2 == 0` vedno neresničen, saj bo `čakam == "da"` `False` in ne `True`.

Bolj običajno bi bilo sicer takole:

````python
čakam = True

for x in s:
    if čakam and x % 2 == 0:
        print(x)
        čakam = False
````

Da `čakam` je lahko res (`True`) ali ne (`False`). V začetku je res (`True`). Pogoj `čakam and x % 2 == 0` bo resničen za prvo sodo število, ki ga bomo našli: `x % 2 == 0` bo `True`, `čakam` pa očitno tudi, torej bomo imeli `True and True`, kar je, seveda, `True`. Prvo sodo število se bo tako izpisalo -- `čakam` pa bo postal `False`. Vsi prihodnji pogoji `čakam and x % 2 == 0` bodo torej neresnični že zato, ker je `čakam` `False`.

Obstaja tudi preprostejša rešitev: zanko `for` lahko, tako kot zanko `while`, prekinemo z `break`.

````python
for x in s:
    if x % 2 == 0:
        print(x)
        break
````

Čim naletimo na sodo število, ga izpišemo in z `break` prekinemo zanko.

Kaj pa tole: če seznam ne vsebuje nobenega sodega števila, bi radi izpisali, da v seznamu ni sodih števil.

Najprej rešimo po starem, s `čakam`.

````python
čakam = True

for x in s:
    if čakam and x % 2 == 0:
        print(x)
        čakam = False

if čakam:
    print("Čakam in še vedno čakam, čeprav je seznama že konec.")
    print("Seznam torej ne vsebuje sodih števil.")
````

````python
s = [1, 3, 5, 7]

čakam = True

for x in s:
    if čakam and x % 2 == 0:
        print(x)
        čakam = False

if čakam:
    print("Čakam in še vedno čakam, čeprav je seznama že konec.")
    print("Seznam torej ne vsebuje sodih števil.")
````

Če je `čakam` po zanki še vedno `True`, ga očitno nismo postavili na `False`. Torej v seznamu ni bilo sodih števil.

Če namesto dodatne spremenljivke uporabimo `break`, je stvar malo bolj zanimiva. Če spodnjega ne boste razumeli, nič hudega. Tem, ki jim gre programiranje boljše od rok, pa le pokažimo.

Takole lahko storimo.

````python
s = [5, 8, 3, 12, 4, 5]

for x in s:
    if x % 2 == 0:
        print(x)
        break
else:
    print("V seznamu ni sodih števil")
````

````python
s = [1, 3, 5, 7]
for x in s:
    if x % 2 == 0:
        print(x)
        break
else:
    print("V seznamu ni sodih števil")
````

Pazite: `else` je poravnan s `for` in se nanaša na `for`, ne na `if`. Kar napišemo v `else`, se zgodi, če se zanka *ne prekine z break*.

Logika je v tem: v zanki `for` pogosto iščemo kakšno reč. Ko jo najdemo, z njo nekaj naredimo (jo izpišemo, izbrišemo, vrnemo, z njo pokličemo določeno funkcijo ...) in nato prekinemo zanko z `break`, saj smo to, kar smo iskali, našli in nima smisla, da bi zanka tekla naprej.

Blok `else` pove, kaj storiti "sicer", namreč če tistega, kar iščemo, nismo našli.

Tako bi lahko preprosto napisali tudi program, ki pove, ali seznam vsebuje soda števila ali ne - ne da bi jih izpisoval.

````python
s = [1, 3, 5, 7]

for x in s:
    if x % 2 == 0:
        print("Seznam vsebuje soda števila.")
        break
else:
    print("V seznamu ni sodih števil")
````
