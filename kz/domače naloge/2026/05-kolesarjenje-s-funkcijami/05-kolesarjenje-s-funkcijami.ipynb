## Rešitev

Prve tri funkcije so samo variacije na to, kar smo v bistvu počeli že prejšnji teden.

Pa ponovimo. Za poskušanje si pripravimo nek seznam `s`.

```python
s = [7, 4, 5, 8, 9, 2, 3]
```

Predstavljajmo si dva seznama - `s` in `s` brez prvega elementa.

```python
print(s)
print(s[1:])
```

Če ju zipnemo skupaj, dobimo pare `(7, 4)`, `(4, 5)`, `(5, 8)` in tako naprej - torej zaporedne elemente seznama `s`. Zanimajo nas njihove razlike, torej

```python
for prej, potem in zip(s, s[1:]):
    print(potem - prej)
```

Pravzaprav jih ne želimo izpisovati, temveč zložiti v seznam:

```python
r = []
for prej, potem in zip(s, s[1:]):
    r.append(potem - prej)

print(r)
```

Zdaj to le spakiramo v funkcijo. Imenovala se bo, kot zahteva naloga `razlike` in argument bomo poimenovali `s`: karkoli bomo funkciji dali kot argument, bo ta videla pod imenom `s`. Sledi zanka, kot jo imamo zgoraj, le namesto da bi dobljeni seznam izpisali, ga bomo vrnili.

```python
def razlike(s):
    r = []
    for prej, potem in zip(s, s[1:]):
        r.append(potem - prej)
    return r
```

Preverimo, ali deluje.

```python
razlike([7, 4, 5, 8, 9, 2, 3])
```

Imenitno. Druga naloga je zdaj preprosta: funkcija dobi seznam višin, pokliče funkcijo `razlike`, da dobi razlike med njimi, potem pa sešteje pozitivne razlike.

```python
def skupna_visina(visine):
    vis_razlike = razlike(visine)
    skupna = 0
    for razlika in vis_razlike:
        if razlika > 0:
            skupna += razlika
    return skupna
```

Deluje?

```python
skupna_visina([306, 330, 303, 322, 333, 353, 350, 330, 352, 306])
```

Odgovor je skladen z navodili (in navodila, tokrat, s pravilnim odgovorom).

Gre tudi malenkost hitreje: spremenljivke `vis_razlike` pravzaprav ne potrebujemo.

```python
def skupna_visina(visine):
    skupna = 0
    for razlika in razlike(visine):
        if razlika > 0:
            skupna += razlika
    return skupna
```

```python
skupna_visina([306, 330, 303, 322, 333, 353, 350, 330, 352, 306])
```

V tretji nalogi gremo istočasno čez višinske razlike, ki jih dobimo z `razlike(visine)` in razdalje, ki jih dobimo z `razlike(oddaljenosti)`, ter delimo istoležne elemente.

```python
def vzponi(visine, oddaljenosti):
    v = []
    for y, x in zip(razlike(visine), razlike(oddaljenosti)):
        v.append(y / x)
    return v
```

V četrti nalogi nas spet zanimajo višinske razlike, torej `razlika(visine)`. Vrnemo prvi negativni element. Če ni bilo nobenega, vrnemo 0.

```python
def prvi_spust(visine):
    for razlika in razlike(visine):
        if razlika < 0:
            return razlika
    return 0
```
