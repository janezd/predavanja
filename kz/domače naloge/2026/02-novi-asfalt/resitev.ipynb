Piše se leto 2026 in tudi v Ljubljani se bližajo volitva za župana.

*Mestna občina Ljubljana* (MOL) zato ugotovi, da je napočil čas za prenovo razpadajoče kolesarske steze ob Večni poti. *Oddelek za gozdarske dejavnosti in motorni promet* po podrobni preučitvi situacije ugotovi, da so zadnje zaplate asfalta dejansko izginile in sklene, da bodo vsako leto asfaltirali tretjino poti. Kolesarji se razveselé, da bo pot asfaltirana že v treh letih, vendar se izkaže, da je to mišljeno malo drugače: vsako leto bodo asfaltirali tretjino **preostale dolžine poti**.

Recimo, da je celotna pot dolga 2700 metrov.

- Prvo leto bodo asfaltirali eno tretjino, to je 2700 / 3 = **900** metrov. Neasfaltiranih bo še 1800 metrov.
- Drugo leto bodo asfaltirali eno tretjino ostanka, torej 1800 / 3 = **600** metrov. Neasfaltiranih bo še 1200 metrov.
- Tretje leto bodo asfaltirali eno tretjino preostanka, torej 1200 / 3 = **400** metrov. Neasfaltiranih bo še 800 metrov.
- ...

Kolesarji - presneta banda nehvaležna! - seveda začnejo robantiti, da tega ne bo nikoli konec in opletati z limitami, [pa Ahilom, pa želvami](https://en.wikipedia.org/wiki/Infinity#Zeno:_Achilles_and_the_tortoise)... MOL jih v svoji neskončni potrpežljivosti pomiri, da bodo metre zaokrožali navzgor:

- ...
- Četrto leto bodo asfaltirali tretjino od 800, to je 267 metrov (266.6666666 zaokroženo navzgor), ostalo bo še 533 metrov.
- Peto leto pride na vrsto 178 metrov (177.666666 zaokroženo navzgor), ostane še 355 metrov.
- Šesto leto dodajo 119 metrov (118.33333333 zaokroženo navzgor), ostane 236 metrov.
- ...

Izkaže se, da se prenova ne bo vlekla v neskončnost, temveč bo končana v borih **19** letih, kar je, kot pripomni MOL, bistveno prej kot nikoli. Zlobno natolcevanje kolesarjev o odnosu MOL do kolesarske infrastrukture je s tem -- kot vedno! -- razkrinkano kot popolnom neresnično.

"Vpišemo dolžino poti" lahko pomeni tudi, da na začetku programa piše, preprosto, (na primer) `dolzina = 2700`.

**Pomoč:** x // 3 vrne rezultat deljenja x z 3, zaokrožen navzdol. Če potrebujemo rezultat, zaokrožen navzgor, k x prišetejemo 2, predem ga delimo s 3, torej `(x + 2) // 3`. ### Rešitev

Beležiti si bo potrebno dve stvari: koliko je še `potrebno_asfaltirati` in koliko bomo asfaltirali `letos`. Pri tem bo `letos` seveda izračunan kot `(potrebno_asfaltirati + 2) // 3`. Program brez zanke bi bil lahko videti takole.

````python
potrebno_asfaltirati = 2700

letos = (potrebno_asfaltirati + 2) // 3
print(letos)
potrebno_asfaltirati -= letos

letos = (potrebno_asfaltirati + 2) // 3
print(letos)
potrebno_asfaltirati -= letos

letos = (potrebno_asfaltirati + 2) // 3
print(letos)
potrebno_asfaltirati -= letos

letos = (potrebno_asfaltirati + 2) // 3
print(letos)
potrebno_asfaltirati -= letos

letos = (potrebno_asfaltirati + 2) // 3
print(letos)
potrebno_asfaltirati -= letos
````

Ker pa smo že dovolj pametni, vemo, da treh vrstic programa ne ponavljamo tako, da jih kopiramo, temveč jih zapremo v zanko `while`. Doklej pa je potrebno to ponavljati? Dokler ni asfaltirano vse, torej dokler `potrebno_asfaltirati > 0`.

````python
potrebno_asfaltirati = 2700

while potrebno_asfaltirati > 0:
    letos = (potrebno_asfaltirati + 2) // 3
    print(letos)
    potrebno_asfaltirati -= letos
````

## Dodatni nalogi (ki pa ne zahtevata dodatnega znanja)

Tidve nalogi nista obvezni; rešujte ju, če želite.

- Izpišite še, koliko let bo trajalo to asfaltiranje (torej: koliko vrstic bo izpisanih).
- Izpišite, koliko metrov bodo asfaltirali v prvih desetih letih in koliko bo še ostalo.

### Rešitev

Leta štejemo tako, da jih povečamo ob, recimo, vsakem izpisu.

````python
potrebno_asfaltirati = 2700
leta = 0

while potrebno_asfaltirati > 0:
    letos = (potrebno_asfaltirati + 2) // 3
    print(letos)
    leta += 1
    potrebno_asfaltirati -= letos

print("Vse skupaj bo trajalo", leta, "let.")
````

Če hočemo reč ustaviti po desetih letih, lahko, recimo, v desetem letu prekinemo zanko z `break`.

````python
potrebno_asfaltirati = 2700
leta = 0

while potrebno_asfaltirati > 0:
    letos = (potrebno_asfaltirati + 2) // 3
    potrebno_asfaltirati -= letos
    print(letos)
    leta += 1
    if leta == 10:
        break
````

Koliko je potrebno še asfaltirati, je zapisano v `potrebno_asfaltirati`. Koliko pa so asfaltirali? `2700 - potrebno_asfaltirati`, vendar smo številko 2700 medtem že izgubili. To bomo rešili tako, da si jo bomo zapomnili v ločeni spremenljivki.

````python
pot = 2700
potrebno_asfaltirati = pot
leta = 0

while potrebno_asfaltirati > 0:
    letos = (potrebno_asfaltirati + 2) // 3
    potrebno_asfaltirati -= letos
    print(letos)
    leta += 1
    if leta == 10:
        break

print("Asfaltirati je potrebno še", potrebno_asfaltirati)
print("Asfaltirali so", pot - potrebno_asfaltirati)
````

Pa če nočemo prekiniti zanke? V tem primeru si, recimo, zapomnimo, koliko je ostalo po desetih letih.

````python
pot = 2700
potrebno_asfaltirati = pot
leta = 0
ostalo_po_10 = 0

while potrebno_asfaltirati > 0:
    letos = (potrebno_asfaltirati + 2) // 3
    potrebno_asfaltirati -= letos
    print(letos)
    leta += 1
    if leta == 10:
        ostalo_po_10 = potrebno_asfaltirati

print("Po desetih letih je ostalo", ostalo_po_10)
print("Asfaltirali so", pot - ostalo_po_10)
````

Odstranili smo torej `break` in pustili zanko steči do konca, vendar smo si zapomnili, kaj bi izpisali, če bi jo končali v tem trenutku.
