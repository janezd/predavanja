## Kolesarski zapiski

Kolesar Jože je šel kolesarit. V začetku je bil na nadmorski višini 306 m in 0 m oddaljen od doma. Čez eno minuto je pogledal na števec in videl, da je na višini 330 m in 442 m oddaljen od doma; najbrž je živel pod hribom in se pošteno zagnal vanj. Čez dve minuti je bil na 303 m (najbrž na drugi strani hriba) in 892 m od doma. Na števec je še naprej pogledal vsako minuto in ker je imel izjeme spomin, si je vse te številke zapomnil, tako da jih je lahko doma shranil v dva seznama v Pythonu.

````python
visine = [306, 330, 303, 322, 333, 353, 350, 330, 352, 306]
oddaljenosti = [   0,  442,  892, 1259, 1584, 1960, 2310, 2717, 3096, 3628]
````

Program naj se začne s tema seznamoma, napisan pa naj bo tako, da bi pravilno deloval tudi z drugačnimi številkami. Torej: v program ne dodajajte kakšnih številk, ki bi jih ročno izračunali iz teh seznamov, ampak naj program sam izračuna vse, kar je potrebno.

### 1. Razdalje

Izpiši razdalje, ki jih je prevozil v posameznih minutah: v prvi minuti je prevozi 442 m, v drugi 450 m, v tretji 367 m, itd. To, očitno izračunaš tako, da od vsake oddaljenosti odšteješ prejšnjo oddaljenost. Na predavanju nismo nikoli uporabili zip na ta način, vendar lahko uganete, kaj vam je storiti. Če ne, pa poglejte v zapiske, kjer se skriva tudi to. (Razni GPTji vam bodo najbrž dali drugačen nasvet, vendar priporočam, da se gibate v okviru tega, kar se učimo in razumemo.)

#### Rešitev

Uporabimo lahko `zip`.

Za ilustracijo si najprej izpišimo seznam `oddaljenosti` in taisti seznam brez prvega elementa.

````python
print(oddaljenosti)
print(oddaljenosti[1:])
````

Če z `zip` poparimo tadva seznama, bomo dobili pare `[0, 442]`, `[442, 892]`, `[892, 1259]` in tako naprej. Poimenujmo elemente teh parov `prej` in `potem`. Program mora izpisati razlike med njima.

````python
for prej, potem in zip(oddaljenosti, oddaljenosti[1:]):
    print(potem - prej)
````

V Pythonovem modulu `itertools` najdemo funkcijo `pairwise`. `pairwise(s)` vrne isto stvar kot `zip(s, s[1:])`. (Zakaj potem zanjo nisem povedal prej? Ker zna `zip` tudi še kaj drugega in ga je praktično poznati. Po drugi strani pa lahko s `pairwise` naredimo pare zaporednih elementov tudi v nekaterih situacijah, ko ne bi mogli uporabiti `zip(s, s[1:])`.) Program lahko torej napišemo tudi tako:

````python
from itertools import pairwise

for prej, potem in pairwise(oddaljenosti):
    print(potem - prej)
````

Samo tisto, kar smo izpisovali v prejšnji nalogi, moramo deliti s 60 in pomnožiti s 3.6.

````python
from itertools import pairwise

for prej, potem in pairwise(oddaljenosti):
    print((potem - prej) / 60 * 3.6)
````

### 3. Vzponi

Izpišite povprečni naklon v procentih (to je, metrih na meter) za vsako minuto. To storimo tako: v prvi minuti se je dvignil za 24 m (330 - 306) in prevozil 442 m, kar pomeni, da je bil naklon 24/442 = 0.0543, kar je 5.43 %. V drugi minuti se je spustil za 27 m (303 - 330) in prevozil 450 m, kar pomeni, da je bil naklon -27/450 = -0.06, kar je -6 %. In tako naprej. Naklon je torej razlika v višini deljena z razdaljo, ki jo je prevozil, pomnožena s 100.

Nasvet: naloge se lahko lotite tako, da najprej izračunate (ne ročno vnesete!!!) seznama razdalj in višinskih razlik. Prvi bo vseboval številke, ki ste jih izpisali v prvem delu (samo dodajajte jih v seznam namesto da bi jih izpisovali!), drugi pa razlike v višini, ki jih izračunate na enak način kot razdalje. Ko imate oba seznama, ustrezno delite in množite njune vrednosti.

Če hočete lepši izpis, lahko uporabite funkcijo round, ki zaokroži podano število.

#### Rešitev

Pa si pripravimo oba seznama, kot so svetovala navodila. Potem se vzporedno sprehodimo po njima in delimo njune elemente.

````python
razdalje = []
for prej, potem in pairwise(oddaljenosti):
    razdalje.append(potem - prej)


vrazlike = []
for prej, potem in pairwise(visine):
    vrazlike.append(potem - prej)

for r, v in zip(razdalje, vrazlike):
    print(round(v / r * 100, 2), "%")
````

Ne spreglejmo: pri pripravi seznamov smo uporabili `pairwise`, v zadnji zanki pa seveda `zip`, saj je bilo potrebno iti istočasno prek elementov dveh seznamov, medtem ko gre `pairwise` prek zaporednih elementov enega seznama.

Če si ne bi poprej pripravili obeh seznamov, bi lahko v enem zamahu sestavljali pare zaporednih elementov obeh seznamov, takole.

````python
for r1, r2, v1, v2 in zip(oddaljenosti, oddaljenosti[1:], visine, visine[1:]):
    print(round((v2 - v1) / (r2 - r1) * 100, 2), "%")
````

Če je zadnja vrstica prezapletena (najmanj, kar lahko rečemo, je, da ni prav pregledna), jo lahko razpišemo.

````python
for r1, r2, v1, v2 in zip(oddaljenosti, oddaljenosti[1:], visine, visine[1:]):
    v = v2 - v1
    r = r2 - r1
    naklon = v / r * 100
    print(round(naklon, 2), "%")
````

### 4. Skupna višina

Izračunajte in izpišite skupno vsoto vzponov. V prvi minuti se je dvignil za 24 m, v drugi spustil za 27 m, v tretji se je dvignil za 19 m, itd. Celoten seznam vzponov in spustov je torej `[24, -27, 19, 11, 20, -3, -20, 22, -46]` (ta seznam naj se izračuna v programu iz podanih seznamov: ne kopirajte ga v program!). Ker nas zanima le vsota vzponov, seštejte vsa pozitivna števila, torej 24 + 19 + 11 + 20 + 22 = 96 m.

#### Rešitev

Spet enaka zanka kot one od prej, le še en pogoj nam je dodati.

````python
vzponi = 0
for prej, potem in pairwise(visine):
    if potem > prej:
        vzponi += potem - prej

print(vzponi)
````

Po drugi strani smo si že prej pripravili seznam višinskih razlik. Lahko ga uporabimo.

````python
vzponi = 0
for razlika in vrazlike:
    if razlika > 0:
        vzponi += razlika

print(vzponi)
````

To nas, končno, privede do ideje, da bi lahko rešili vse tri naloge z vnaprej pripravljenimi seznami.

### Vse skupaj, še enkrat

````python
visine = [306, 330, 303, 322, 333, 353, 350, 330, 352, 306]
oddaljenosti = [   0,  442,  892, 1259, 1584, 1960, 2310, 2717, 3096, 3628]

razdalje = []
for prej, potem in pairwise(oddaljenosti):
    razdalje.append(potem - prej)


vrazlike = []
for prej, potem in pairwise(visine):
    vrazlike.append(potem - prej)

print("1. naloga")
for r in razdalje:
    print(r)
print()

print("2. naloga")
for r in razdalje:
    print(r / 60 * 3.6)
print()

print("3. naloga")
for r, v in zip(razdalje, vrazlike):
    print(round(v / r * 100, 2), "%")
print()

print("4. naloga")
vzponi = 0
for razlika in vrazlike:
    if razlika > 0:
        vzponi += razlika

print(vzponi)
````
