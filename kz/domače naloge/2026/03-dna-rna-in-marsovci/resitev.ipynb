## Obvezna naloga

DNK vsebuje štiri različne baze: A, T, G in C. Tudi RNK pa vsebuje štiri različne baze: A, U, G in C -- namesto timina ima uracil. Napišite program, ki bo preveril, ali je dani niz DNK ali RNK.

Na začetek programa si skopirajte naslednje nize.

````python
a = "ATGGAGCGTGGTTGCGAGCGATAATCGACTAGCTAC"
b = "ACGUACAGCGAUCAGCUAACGAUGCAUGCAGGUACG"
c = "ACGUACAGCGTCAGCUAACGAUGCAUGCAGGUACG"
d = "AACGACGGGACAGCAGCGAGCAGCAGCCGACG"
````

Vaš program bo vseboval zanko, ki bo šla prek vseh črk v posamičnem nizu. Program boste spreminjali tako, da boste enkrat pisali `for x in a`, drugič `for x in b` itd.

- Če niz vsebuje kakšen T in nobenega U (primer `a`), naj program izpiše "DNK".
- Če niz vsebuje kakšen U in nobenega T (primer `b`), naj program izpiše "RNK".
- Če niz vsebuje tako T kot U (primer `c`), naj program izpiše "Marsovec".
- Če niz ne vsebuje niti T niti U (primer `d`), naj program prav tako izpiše "Marsovec".

## Rešitev

Očitno se bomo sprehodili čez niz, recimo `a` in si, če slučajno naletimo na T, to zapomnili. Pa če naletimo na U, tudi.

````python
imamo_t = False
imamo_u = False

for baza in a:
    if baza == "T":
        imamo_t = True
    if baza == "U":
        imamo_u = True
    
````

Oba, `imamo_t` in `imamo_u`, v začetku nastavimo na `False`; če ju kasneje spremenimo na `True`, sta pač `True`. Med DNK, RNK in marsovci zdaj ločimo s par pogoji.

````python
if imamo_t and not imamo_u:
    print("DNK")
elif imamo_u and not imamo_t:
    print("RNK")
else:
    print("Marsovec")
````

Tule je pomembno, da uporabimo `elif`, sicer bi se to, kar piše v zadnje `else`, zgodilo tudi pri tistih, ki ustrezajo prvemu pogoju; pri vsakem DNK bi se poleg tega izpisalo še Marsovec.

Obstaja tudi krajša rešitev: z operatorjem `in` lahko preverimo, ali niz vsebuje določen podniz:

````python
"T" in a
````

````python
"U" in a
````

Če želimo preveriti, da "U" ni v nizu `a`, lahko uporabimo operator `not`,

````python
not "U" in a
````

Da se stvari lepše berejo obstaja tudi operator `not in`, ki preveri, da niz ne vsebuje določenega podniza.

````python
"U" not in a
````

(Ne, da je zelo pomembno, ampak vseeno povem: `not "U" in a` najprej izračuna `"U" in a` in potem zanika rezultat -- v bistvu je to `not ("U" in a)`. `not in` pa je dve besedi dolg operator -- v sintaktičnem smislu je to en "token", ena "enota".)

Skratka. Nalogo lahko rešimo kar brez zanke, oziroma se "zanka" zgodi znotraj "in".

````python
if "T" in a and "U" not in a:
    print("DNK")
elif "U" in a and "T" not in a:
    print("RNK")
else:
    print("Marsovec")
````

## Rešitev dodatne naloge

Spremljati moramo dolžino trenutnega zaporedja in hraniti dolžino najdaljšega zaporedja, kar smo jih videli.

````python
e = "AATGCGAAAATGTGAAAAA"

najdaljse = 0
to = 0
for baza in e:
    if baza == "A":
        to += 1
        if to > najdaljse:
            najdaljse = to
    else:
        to = 0

print("Najdaljše zaporedje A-jev je dolgo:", najdaljse)
````

Če naletimo na A, povečamo dolžino trenutnega zaporedja (`to`) in preverimo, če je `to` zdaj daljše kot `najdaljse`.

Če trenutna baza ni A, pa postavimo dolžino trenutnega zaporedja na `0`.
