# Izdatki

Nekdo si zapisuje svoje izdatke v datoteko. Vsak izdatek je v svoji vrstici. Prva črka pove, za kakšen izdatek gre: lahko je `H` (hrana), `B` (bivanje) ali `O` (ostalo). Sledi še znesek izdatka.

Datoteka je torej lahko videti takole:

```
H23
H5
B30
H8.25
O2
O4
B22
```

Napišite dve funkciji.

- `razdeli(s)` prejme niz, ki je takšne oblike kot ena vrstica datoteke. Vrniti mora seznam z dvema elementoma: prvi element je črka, ki označuje vrsto izdatka, drugi element pa je znesek izdatka kot `float`. Na primer, `razdeli("H23.25")` vrne `["H", 23.25]`.

    Funkcija naj deluje tudi, če se vrstica začne s kako drugo črko. Tako naj `razdeli("X100")` vrne `["X", 100.0]`.

- `izdatki(ime_datoteke)` prejme ime datoteke, ki vsebuje izdatke v zgoraj opisani obliki. Funkcija naj prebere datoteko in vrne seznam s tremi elementi: skupni stroški hrane, bivanja in ostalih izdatkov. Na primer, če datoteka vsebuje zgoraj prikazane vrstice, naj funkcija vrne `[36.25, 52.0, 6.0]`.

## Rešitev

V prvi funkciji ponavljamo indeksiranje. Pa, seveda, pisanje funkcij.

````python
def razdeli(vrstica):
    tip = vrstica[0]
    znesek = float(vrstica[1:])
    return [tip, znesek]
````

Če komu ni pretežko gledati malo več oklepajev v isti vrstici, pa ne sestavlja dodatnih spremenljivk.

````python
def razdeli(vrstica):
    return [vrstica[0], float(vrstica[1:])]
````

V drugi funkciji pripravimo spremenljivke `h`, `b` in `o` ter seštevamo izdatke vanje.

````python
def izdatki(ime_datoteke):
    hrana = 0
    bivanje = 0
    ostalo = 0
    for vrstica in open(ime_datoteke):
        tip, znesek = razdeli(vrstica)
        if tip == "H":
            hrana += znesek
        if tip == "B":
            bivanje += znesek
        if tip == "O":
            ostalo += znesek
    return [hrana, bivanje, ostalo]
````
