## Čas kolesarjenja

Kolesar, ki se je namenil prevoziti razdaljo $s$ km s hitrostjo $v$ km/h, bo za svoj podvig potreboval čas $t = s / v$ ur. Da mu ne bo potrebno računati ročno, sestavite program, ki mu vnesemo razdaljo in hitrost, nato pa izračuna in izpiše potreben čas za prevoz razdalje. Čas naj bo izpisan kar v urah; 3.8 ure pač pomeni 3 ure in 48 minut.

Program naj nato izpiše še naslednje.

- Če je čas daljši od ene ure, mora izpisati, naj kolesar ne pozabi pijače. (To naj se zgodi vedno, ko je čas daljši od ene ure, ne glede na ostale izpise.)
- Če je čas daljši od dveh ur, vendar krajši od štirih ur, naj izpiše, da naj ne pozabi vzeti hrane.
- Če je čas daljši od štirih ur, naj izpiše, naj ne pozabi iti nekam na pizzo.
Program oddajte kot datoteko .py ali .ipynb.

## Rešitev

```python
razdalja = int(input("Razdalja: "))
hitrost = int(input("Hitrost: "))

cas = razdalja / hitrost

print("Čas kolesarjenja:", cas, "ur.")

if cas > 1:
    print("Ne pozabi pijače")

if 2 < cas < 4:
    print("Ne pozabi vzeti hrane.")
elif cas >= 4:
    print("Ne pozabi iti nekam na pizzo")
```

Naloga je, očitno, naloga iz pogojnih stavkov in pravilnega postavljanja `if`-ov.

Zadnji `elif` bi lahko bil tudi samo `if` (razmislite, zakaj), ne smemo pa pisati le

```python
if 2 < cas < 4:
    print("Ne pozabi vzeti hrane.")
else:
    print("Ne pozabi iti nekam na pizzo")
```

V tem primeru bi namreč poslal na pizzo tudi tiste, ki so kolesarili manj kot dve uri in ne le tistih, ki bodo na kolesu vsaj štiri.

Končno, poglejmo še pogoja `cas < 4` in `cas >= 4`. Vsaj v enem mora biti enačaj, sicer tisti, ki so kolesarijo natančno štiri ure, ne bodo niti opomnjeni na hrano niti poslani na pizzo.
