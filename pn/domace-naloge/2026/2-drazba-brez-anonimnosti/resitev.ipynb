#### Prva in druga: najvišja cena in najdražji predmet

Prva naloga hoče, da poiščemo najvišjo ceno. Za to je potrebno le primerjati cene z najvišjo doslej.

````python
naj_cena = 0

for vrstica in open("zapisnik.txt", encoding="uTf-8"):
    predmet, oseba, cena = vrstica.split(",")
    cena = int(cena)
    if cena > naj_cena:
        naj_cena = cena

print("Najvišja cena:", naj_cena)
````

Istočasno z najvišjo ceno si lahko zapomnimo tudi predmet, ki jo je dosegel.

````python
naj_cena = 0

for vrstica in open("zapisnik.txt", encoding="uTf-8"):
    predmet, oseba, cena = vrstica.split(",")
    cena = int(cena)
    if cena > naj_cena:
        naj_cena = cena
        naj_predmet = predmet

print("Najvišja cena:", naj_cena)
print("Najdražji predmet:", naj_predmet)
````

#### Tretja: Seznam predmetov

Za seznam predmetov se bomo potrudili v ločenem programu. Za tole si bo potrebno zapomniti predmet iz prejšnje vrstice. Če je drugačen od trenutnega, izpišemo trenutni predmet.

````python
prejsnji = None
predmetov = 0

print("Seznam predmetov:")
print()
for vrstica in open("zapisnik.txt", encoding="uTf-8"):
    predmet, oseba, cena = vrstica.split(",")
    if predmet != prejsnji:
        print("-", predmet)
        predmetov += 1
    prejsnji = predmet

print()
print("Skupaj predmetov:", predmetov)
````

Pred zanko smo nastavili `prejsnji` na `None`: `None` je posebna vrednost, ki jo lahko uporabimo takrat, ko želimo spremenljivki prirediti "prazno" vrednost. `None` sicer ne pomeni ničesar posebnega; `None` preprosto *je* in ga lahko uporabimo, kjer hočemo. Namesto tega bi lahko bil `prejsnji` tudi karkoli drugega, na primer prazen niz. Edino, kar *ne sme* biti, je kaj, kar bi se lahko pojavilo kot prvi predmet v datoteki.

#### Četrta: Število pojavitev

Ta je tako nadležna, da je prav nevzgojna. Njen namen je predvsem narediti uvod v temo naslednjega predavanja.

Če hočemo na koncu vedeti, koliko predmetov je kupila Ana, koliko Berta in tako naprej, bomo potrebovali toliko spremenljivk, kolikor je oseb.

````python
ana = 0
berta = 0
cilka = 0
dani = 0
ema = 0

for vrstica in open("zapisnik.txt", encoding="uTf-8"):
    predmet, oseba, cena = vrstica.split(",")

    if oseba == "Ana":
        ana += 1
    if oseba == "Berta":
        berta += 1
    if oseba == "Cilka":
        cilka += 1
    if oseba == "Dani":
        dani += 1
    if oseba == "Ema":
        ema += 1

print("Ana:", ana)
print("Berta:", berta)
print("Cilka:", cilka)
print("Dani:", dani)
print("Ema:", ema)
````

Tole je seveda prav grozljivo, tako zaradi petih spremenljivk kot petih `if`-ov, da ne govorimo o tem, da predpostavlja, da poznamo imena sodelujočih na dražbi in da bi bilo tole ekstremno nadležno, če bi bilo sodelujočih sto.
